```python
data = pd.read_csv("student_scores.csv")  # Read the CSV file into a Pandas DataFrame named data.
print(data)  # Display the complete DataFrame.
```

```python
print(data.head())  # Display the first five rows of the dataset.
print(data.shape)  # Display the number of rows and columns.
print(data.columns)  # Display all column names.
print(data.dtypes)  # Display the data type of each column.
```

```python
math_scores = data["math"]  # Select the math column from the DataFrame.
science_scores = data["science"]  # Select the science column from the DataFrame.
print(math_scores)  # Display all math scores.
print(science_scores)  # Display all science scores.
```

```python
average_math = data["math"].mean()  # Calculate the average math score.
average_science = data["science"].mean()  # Calculate the average science score.
highest_math = data["math"].max()  # Find the highest math score.
print(average_math)  # Display the average math score.
print(average_science)  # Display the average science score.
print(highest_math)  # Display the highest math score.
```

```python
most_hours = data["study_hours"].max()  # Find the largest number of study hours.
student_index = data["study_hours"].idxmax()  # Find the row index containing the largest study-hours value.
student_name = data.loc[student_index, "student"]  # Use the row index to find the student's name.
print(most_hours)  # Display the largest number of study hours.
print(student_name)  # Display the student who studied the most.
```

```python
data["average_score"] = (data["math"] + data["science"]) / 2  # Calculate each student's average score.
print(data)  # Display the updated DataFrame with the new column.
```

```python
plt.bar(data["student"], data["average_score"])  # Create a bar chart using student names and average scores.
plt.title("Average Score by Student")  # Add a title to the chart.
plt.xlabel("Student")  # Label the horizontal axis.
plt.ylabel("Average Score")  # Label the vertical axis.
plt.xticks(rotation=45)  # Rotate student names so they are easier to read.
plt.tight_layout()  # Adjust spacing so labels fit properly.
plt.show()  # Display the finished chart.
```

```python
data.to_csv("student_scores_with_average.csv", index=False)  # Save the updated DataFrame to a new CSV file without row numbers.
print("File saved successfully.")  # Confirm that the new file was created.
```
